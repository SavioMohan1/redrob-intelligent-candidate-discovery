## 4. Preview ranked candidates

```python
import csv
from pathlib import Path

csv_path = Path('demo/demo_submission.csv')
with csv_path.open(encoding='utf-8', newline='') as f:
    rows = list(csv.DictReader(f))

for row in rows[:10]:
    print(f"#{row['rank']:>3} {row['candidate_id']} score={row['score']} - {row['reasoning']}")
```
